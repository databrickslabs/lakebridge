# Switch post-transpile schema remapping

CSV-based table namespace and column renames for SQL under `output_dir`. Optional `default_catalog` / `default_schema` prepend `USE CATALOG` / `USE SCHEMA` before each file.

`remap_max_workers` (default 16) controls concurrent per-file I/O; set to 1 to force sequential processing.

The `apply_schema_remap` widget defaults to **true** for ad-hoc runs. The Switch job can override with **false** to skip remapping quickly.

```python
dbutils.widgets.text("output_dir", "")
dbutils.widgets.text("namespace_remap_csv_path", "")
dbutils.widgets.text("column_remap_csv_path", "")
dbutils.widgets.text("default_catalog", "")
dbutils.widgets.text("default_schema", "")
dbutils.widgets.text("apply_schema_remap", "true")
dbutils.widgets.text("remap_max_workers", "16")

from databricks.labs.lakebridge.transpiler.switch_schema_remap import run

summary = run(
    dbutils.widgets.get("output_dir"),
    dbutils.widgets.get("namespace_remap_csv_path"),
    dbutils.widgets.get("column_remap_csv_path"),
    dbutils.widgets.get("default_catalog"),
    dbutils.widgets.get("default_schema"),
    dbutils,
    apply_schema_remap=dbutils.widgets.get("apply_schema_remap"),
    max_workers=dbutils.widgets.get("remap_max_workers"),
)
print(summary)
```
